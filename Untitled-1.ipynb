Method for reading text from color images applies:
Greyscale to acentuate the contrast
INTER_CUBIC - defines how pixels are enlarged, enlarging text which is beneficial for OCR
Scaling factors fx=fy=2 is apply to be applied to height and width


```python
def read_text_from_color_image(img_path):

    img = cv2.imread(img_path,cv2.IMREAD_GRAYSCALE)
    
    # Apply gray scale and resize for better visibility
    #
    gray_img = cv2.resize(
        #cv2.cvtColor(img, cv2.COLOR_BGR2GRAY),
        img,
        None,
        fx=2,
        fy=2,
        interpolation=cv2.INTER_CUBIC)
  
    # Blur to reduce background noise
    blur = cv2.GaussianBlur(gray_img, (3, 3), 0)
    
    # Apply Threshold: keep bright white text
   # _, thresh = cv2.threshold(blur, 180, 255, cv2.THRESH_BINARY)
    _, thresh = cv2.threshold(
    gray_img,
    180,
    255,
    cv2.THRESH_BINARY
)
#adaptive threshold made things worse
#     thresh = cv2.adaptiveThreshold(
#     blur,
#     255,
#     cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
#     cv2.THRESH_BINARY,
#     31,
#     11
# )
    # 5. Optional: clean small specks
    # GaussianBlur(image, kernel_size, sigma) Larger values = stronger blur. Too much blur destroys letters.
    # Rules:
    # must be odd numbers
    # width and height usually equal
    #Controls Gaussian spread.
    # 0 means: OpenCV automatically chooses sigma from kernel size
    kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (3, 3))

    clean_img = cv2.morphologyEx(thresh, cv2.MORPH_OPEN, kernel)
    #inverted_img = cv2.bitwise_not(clean_img)

    return clean_img
```

```python
path = "docs/text_on_img.png"
clean_img = read_text_from_color_image(path)
display_img(clean_img)
```

```python
def invert_image_colors(clean_img):
    inverted_img = cv2.bitwise_not(clean_img)
    return inverted_img
```

```python
import pytesseract
from PIL import Image

def extract_text_from_img_white(img1, info = ""):
    
    #whitelisting letters
    config = "--psm 6 -c tessedit_char_whitelist=ABCDEFGHIJKLMNOPQRSTUVWXYZabcdefghijklmnopqrstuvwxyz"
   
    text = pytesseract.image_to_string(img1, config=config)
  
    print(info,repr(text))
   
```

```python

def extract_text_from_img_black(img1, info = ""):
    
    #whitelisting letters
    config =  "--psm 6 -c tessedit_char_blacklist=0123456789"
   
    text = pytesseract.image_to_string(img1, config=config)
  
    print(info,repr(text))
```

```python
def extract_text_from_img_cleanup(img, info):
    config = (
    "--psm 6 "
    "-c tessedit_char_blacklist=|[]{}<>~"
)
    text = pytesseract.image_to_string(img, config=config)
  
    print(info,repr(text))
```

```python

display_img(original_image)
extract_text_from_img_white(original_image , "Initial text: ")

display_img(clean_img)
extract_text_from_img_white(clean_img , "Clean Applied Black and White text: ")

invert_img = invert_image_colors(clean_img)
display_img(invert_img)
extract_text_from_img_black(invert_img , "Clean + Inverted Black & White: ")

extract_text_from_img_cleanup(invert_img , "Cleanup config: ")

```
